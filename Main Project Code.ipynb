```python
import pygame
import sys

pygame.init()
screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Game Start Screen")
font = pygame.font.SysFont(None, 60)

# Colors
WHITE = (255, 255, 255)
GRAY = (100, 100, 100)
DARK_GRAY = (50, 50, 50)
YELLOW = (255, 223, 0)

# Button setup
start_button = pygame.Rect(300, 200, 200, 60)
quit_button = pygame.Rect(300, 300, 200, 60)

def draw_button(rect, text):
    pygame.draw.rect(screen, GRAY, rect)
    pygame.draw.rect(screen, DARK_GRAY, rect, 3)
    label = font.render(text, True, WHITE)
    label_rect = label.get_rect(center=rect.center)
    screen.blit(label, label_rect)

def draw_background():
    # Sky
    screen.fill((135, 206, 235))  # Clear blue sky

    # Grass
    pygame.draw.rect(screen, (34, 139, 34), (0, 400, 800, 200))  # Green grass

    # Cloud (fluffy with overlapping ellipses)
    pygame.draw.ellipse(screen, WHITE, (150, 100, 120, 60))
    pygame.draw.ellipse(screen, WHITE, (200, 80, 120, 60))
    pygame.draw.ellipse(screen, WHITE, (250, 100, 120, 60))

    # Sun (top-right corner)
    pygame.draw.circle(screen, YELLOW, (700, 100), 60)

def start_screen():
    while True:
        try:
            # Draw background first
            draw_background()

            # Draw buttons on top
            draw_button(start_button, "Start")
            draw_button(quit_button, "Quit Game")
    
            for event in pygame.event.get():
                if event.type == pygame.QUIT:
                    pygame.quit()
                    sys.exit()
                elif event.type == pygame.MOUSEBUTTONDOWN:
                    if start_button.collidepoint(event.pos):
                        return  # Proceed to game loop
                    elif quit_button.collidepoint(event.pos):
                        pygame.quit()
                        sys.exit()
    
            pygame.display.flip()
        except Exception as e:
            print(e)

# Run the start screen
start_screen()

# Continue with your game loop here...
```
